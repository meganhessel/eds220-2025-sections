2) Date columns to datetime 
3) Index the start date column 

## Cleaning data 

```python
# Change column names 
hurricane.columns = hurricane.columns.str.replace(".", "_")
```

```python
# Change date columh type 
hurricane.start_date = pd.to_datetime(hurricane.start_date)
hurricane.end_date = pd.to_datetime(hurricane.end_date)
```

```python
# Creating a Duration column 
hurricane['duration'] = hurricane['end_date'] - hurricane['start_date']
```

```python
# Change index 
hurricane.set_index('start_date')
```

```python
# How many hurricanes with Saffir-Simpson category 5 have been registered and what was their duration?

hurricane_cat5 = hurricane[hurricane['ss'] == 5]

# Print statements !!
# F statement 
print(f"There are '{hurricane_cat5.shape[0]}' cat 5 hurricanes") # With .shape[0]
print(f"There are '{len(hurricane_cat5)}' cat 5 hurricanes") # With len()

# No f statement 
print("There are", hurricane_cat5.shape[0], "cat 5 hurricanes")
print("There are", len(hurricane_cat5), "cat 5 hurricanes")
```

```python
plt.scatter(hurricane.index, hurricane.ss,
           color = '#485CE0', 
           edgecolors = "white", 
           alpha = 0.7)
plt.yticks([1,2,3,4,5])  # Change the ticks 
plt.xlabel('Years') 
plt.ylabel('Hurricane Category')
plt.title('Hurricanes Across the Yucatan Peninsula 1851-2000')
plt.show()
```
